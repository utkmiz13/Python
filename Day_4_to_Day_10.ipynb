Explicit typecasting:

The conversion of one data type into another data type, done via developer or programmer's

```python
string = "15"
number = 7
string_number = int(string)
sum= number + string_number
print("The Sum of both the numbers is: ", sum)
```

Implicit typecasting

One data type is converted into other by the Python interpreter itself (automatically). This is called, implicit typecasting in python.



```python
# Python automatically converts
# a to int
a = 7
print(type(a))

# Python automatically converts b to float
b = 3.0
print(type(b))

# Python automatically converts c to float as it is a float addition
c = a + b
print(c)
print(type(c))
```
